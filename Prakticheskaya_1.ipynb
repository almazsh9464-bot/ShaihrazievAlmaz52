## Задание 1. Перевод строки в число и обратно

```python
# 1. Перевод строки в число и обратно, вывод типов данных
text = input("Введите числовое значение: ")
print("Исходное значение:", text)
print("Тип исходного значения:", type(text))

number = float(text)
print("Число:", number)
print("Тип числа:", type(number))

text_again = str(number)
print("Снова строка:", text_again)
print("Тип строки:", type(text_again))
```

## Задание 2. Сделать число отрицательным

```python
# 2. Сделать число отрицательным, если оно положительное
number = float(input("Введите число: "))

if number > 0:
    number = -number

print("Результат:", number)
```

## Задание 3. Посчитать оставшихся пассажиров

```python
# 3. Посчитать, сколько пассажиров не поместится в автобус
capacity = int(input("Вместимость автобуса: "))
on_bus = int(input("Сколько пассажиров уже в автобусе: "))
waiting = int(input("Сколько пассажиров ждут: "))

left_waiting = on_bus + waiting - capacity
if left_waiting < 0:
    left_waiting = 0

print("Осталось ждать:", left_waiting)
```
