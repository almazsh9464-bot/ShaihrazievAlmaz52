## Задание 1. Средний балл студентов и обработка ошибок

```python
# Практическая 5
import json


def create_report(input_file, output_file):
    try:
        with open(input_file, "r", encoding="utf-8") as file:
            students = json.load(file)

        if not isinstance(students, list):
            raise TypeError("В файле должен быть список студентов")

        student_results = []
        all_grades = []

        for student in students:
            if not isinstance(student, dict):
                raise TypeError("Каждый студент должен быть словарём")

            name = student["name"]
            grades = student["grades"]

            if not isinstance(name, str):
                raise TypeError("Имя студента должно быть строкой")
            if not isinstance(grades, list):
                raise TypeError("Оценки должны быть списком")
            if len(grades) == 0:
                raise ValueError("Список оценок не должен быть пустым")

            for grade in grades:
                if not isinstance(grade, (int, float)) or isinstance(grade, bool):
                    raise TypeError("Оценки должны быть числами")

            average = sum(grades) / len(grades)
            student_results.append({
                "name": name,
                "average_grade": average
            })
            all_grades.extend(grades)

        if len(all_grades) == 0:
            raise ValueError("В файле нет оценок")

        report = {
            "students": student_results,
            "overall_average": sum(all_grades) / len(all_grades)
        }

        with open(output_file, "w", encoding="utf-8") as file:
            json.dump(report, file, ensure_ascii=False, indent=2)

        print("Отчёт сохранён в", output_file)

    except FileNotFoundError:
        print("Файл не найден:", input_file)
    except json.JSONDecodeError:
        print("Файл содержит некорректный JSON")
    except KeyError as error:
        print("В данных отсутствует ключ:", error)
    except (TypeError, ValueError) as error:
        print("Ошибка в данных:", error)


create_report("students.json", "report.json")
```
