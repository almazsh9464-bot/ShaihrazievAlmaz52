## Задание 1. Нужно ли ставить будильник

```python
# 1. Нужно ли ставить будильник
employed = input("Вы трудоустроены? (да/нет): ").lower() == "да"
vacation = input("Вы в отпуске? (да/нет): ").lower() == "да"

need_alarm = employed and not vacation
print(need_alarm)
```

## Задание 2. Выполнить математическую операцию

```python
# 2. Выполнить математическую операцию
operation = input("Введите операцию (+, -, *, /): ")
value_1 = int(input("Введите первое целое число: "))
value_2 = int(input("Введите второе целое число: "))

if operation == "+":
    print(value_1 + value_2)
elif operation == "-":
    print(value_1 - value_2)
elif operation == "*":
    print(value_1 * value_2)
elif operation == "/":
    if value_2 != 0:
        print(value_1 / value_2)
    else:
        print("На ноль делить нельзя")
else:
    print("Неизвестная операция")
```

## Задание 3. Сумма чисел между ними

```python
# 3. Сумма чисел между двумя числами включительно
first = int(input("Введите первое число: "))
second = int(input("Введите второе число: "))

start = min(first, second)
end = max(first, second)
total = 0

for number in range(start, end + 1):
    total += number

print("Сумма:", total)
```

## Задание 4. Определить квартал и сезон

```python
# 4. Определить квартал и сезон по номеру месяца
month = int(input("Введите номер месяца (1–12): "))

if 1 <= month <= 12:
    quarter = (month - 1) // 3 + 1

    if month == 12 or month <= 2:
        season = "зима"
    elif month <= 5:
        season = "весна"
    elif month <= 8:
        season = "лето"
    else:
        season = "осень"

    print("Квартал:", quarter)
    print("Сезон:", season)
else:
    print("Такого месяца нет")
```

## Задание 5. Средний балл и оценка буквой

```python
# 5. Средний балл и оценка буквой
total_score = 0
scores_count = 0

while True:
    score = float(input("Введите оценку (для завершения введите -1): "))

    if score == -1:
        break

    total_score += score
    scores_count += 1

if scores_count == 0:
    print("Оценок нет")
else:
    average = total_score / scores_count
    print("Средний балл:", average)

    if average >= 90:
        letter = "A"
    elif average >= 80:
        letter = "B"
    elif average >= 70:
        letter = "C"
    elif average >= 60:
        letter = "D"
    else:
        letter = "F"

    print("Оценка:", letter)
```
