## Задание 1. Средний возраст n человек
Запустите ячейку и введите количество людей, затем возраст каждого человека.

```python
n = int(input("Сколько человек? "))
total_age = 0

for i in range(n):
    age = int(input(f"Возраст человека {i + 1}: "))
    total_age += age

if n > 0:
    print("Средний возраст:", total_age / n)
```

## Задание 2. Определить цвет по HEX-коду
Введите шестизначный код, например a0351f.

```python
color_code = input("Введите HEX-код цвета: ").strip().lower()

if color_code.startswith("#"):
    color_code = color_code[1:]

if color_code == "000000":
    print("черный")
elif color_code == "ffffff":
    print("белый")
else:
    red = int(color_code[0:2], 16)
    green = int(color_code[2:4], 16)
    blue = int(color_code[4:6], 16)

    if red == green == blue:
        print("серый")
    elif red >= green and red >= blue:
        print("красный")
    elif green >= red and green >= blue:
        print("зеленый")
    else:
        print("синий")
```
