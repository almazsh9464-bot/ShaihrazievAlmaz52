## Задание 1. Класс студента

```python
# Практическая 6

class Student:
    def __init__(self, name, group, tel_number):
        self.name = name
        self.group = group
        self.grades = []
        self.__tel_number = tel_number

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        return (
            f"Имя: {self.name}, группа: {self.group}, "
            f"оценки: {self.grades}, средний балл: {self.get_average():.2f}, "
            f"телефон: {self.__tel_number}"
        )


student_1 = Student("Анна", "36-64", "+79990000001")
student_2 = Student("Иван", "36-64", "+79990000002")
student_3 = Student("Мария", "36-65", "+79990000003")

student_1.add_grade(5)
student_1.add_grade(4)
student_1.add_grade(5)

student_2.add_grade(4)
student_2.add_grade(3)
student_2.add_grade(5)

student_3.add_grade(5)
student_3.add_grade(5)
student_3.add_grade(4)

print(student_1.info())
print(student_2.info())
print(student_3.info())
```
