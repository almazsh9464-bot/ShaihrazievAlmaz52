## Задание 1. Функции для операций и месяца

```python
# 1. Задание 2 из Практической 2 — операция как функция
def calculate(operation, value_1, value_2):
    if operation == "+":
        return value_1 + value_2
    elif operation == "-":
        return value_1 - value_2
    elif operation == "*":
        return value_1 * value_2
    elif operation == "/":
        if value_2 == 0:
            return "На ноль делить нельзя"
        return value_1 / value_2
    return "Неизвестная операция"


operation = input("Введите операцию (+, -, *, /): ")
value_1 = int(input("Введите первое число: "))
value_2 = int(input("Введите второе число: "))
print(calculate(operation, value_1, value_2))


# Задание 4 из Практической 2 — квартал и сезон как функция
def get_quarter_and_season(month):
    if month < 1 or month > 12:
        return "Неверный номер месяца"

    quarter = (month - 1) // 3 + 1

    if month == 12 or month <= 2:
        season = "зима"
    elif month <= 5:
        season = "весна"
    elif month <= 8:
        season = "лето"
    else:
        season = "осень"

    return quarter, season


month = int(input("Введите номер месяца: "))
print(get_quarter_and_season(month))
```

## Задание 2. Посчитать выбранные символы

```python
# 2. Посчитать указанные символы из словаря
def count_selected_characters(sentence, character_counts):
    for character in sentence.lower():
        if character in character_counts:
            character_counts[character] += 1
    return character_counts


sentence = input("Введите предложение: ")
character_counts = {"а": 0, "о": 0, "и": 0}
print(count_selected_characters(sentence, character_counts))
```

## Задание 3. Посчитать символы от m до z

```python
# 3. Посчитать символы от m до z
def count_from_m_to_z(text):
    count = 0
    for character in text:
        if "m" <= character <= "z":
            count += 1
    return count


text = input("Введите строку из букв a-z: ")
print("Количество символов от m до z:", count_from_m_to_z(text))
```
