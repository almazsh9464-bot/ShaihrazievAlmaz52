## Задание 1. Найти строку в списке

```python
# 1. Найти строку в списке и вывести её индекс
words = ["яблоко", "банан", "груша", "апельсин"]
search_word = input("Какое слово найти? ")

if search_word in words:
    print("Индекс:", words.index(search_word))
else:
    print("Такой строки в списке нет")
```

## Задание 2. Удалить дубликаты

```python
# 2. Удалить дубликаты, сохраняя порядок
items = [1, 2, 2, 3, 1, 4, 3]
unique_items = []

for item in items:
    if item not in unique_items:
        unique_items.append(item)

print("Список без повторов:", unique_items)
```

## Задание 3. Посчитать повторы элементов

```python
# 3. Посчитать повторы элементов с помощью словаря
items = ["яблоко", "банан", "яблоко", "груша", "банан", "яблоко"]
counts = {}

for item in items:
    if item in counts:
        counts[item] += 1
    else:
        counts[item] = 1

print("Количество повторений:", counts)
```

## Задание 4. Удалить минимальное число

```python
# 4. Удалить первое минимальное число, сохранив порядок остальных
numbers = [7, 3, 8, 2, 5, 2, 9]
minimum = min(numbers)
result = []
removed = False

for number in numbers:
    if number == minimum and not removed:
        removed = True
    else:
        result.append(number)

print("Результат:", result)
```
